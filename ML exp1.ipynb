```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x=np.array(eval(input()))
y=np.array(eval(input()))
```

```python
x_mean=np.mean(x)
y_mean=np.mean(y)
```

```python
num=0
denom=0
```

```python
for i in range(len(x)):
    num+=(x[i] -x_mean)*(y[i]-y_mean)
    denom+=(x[i] -x_mean)**2
```

```python
m=num/denom
```

```python
b=y_mean-m*x_mean
print(m,b)
```

```python
y_predicted=m*x+b
print(y_predicted)
```

```python
plt.scatter(x,y)
plt.plot(x,y_predicted,color='red')
plt.show()
```
